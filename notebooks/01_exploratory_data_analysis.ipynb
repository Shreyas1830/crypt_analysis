# Crypto Market - Exploratory Data Analysis
Loads the cleaned/feature-engineered dataset produced by the pipeline and walks through summary stats, distributions, correlations, and time-series diagnostics.

```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))

import pandas as pd
from src.database import Database
from src.cleaning import run_cleaning_pipeline, missing_value_report
from src.feature_engineering import build_feature_set

db = Database()
raw = db.read_all()
clean_df, report = run_cleaning_pipeline(raw)
df = build_feature_set(clean_df)
df.tail()
```

```python
# Summary statistics
df[["price", "market_cap", "volume_24h", "percent_change_24h"]].describe()
```

```python
# Skewness & kurtosis
df[["price", "market_cap", "volume_24h"]].agg(['skew', 'kurt'])
```

```python
# Missing value report
missing_value_report(raw)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df[["price", "market_cap", "volume_24h", "percent_change_24h", "circulating_supply"]].corr()
sns.heatmap(corr, annot=True, cmap="RdBu", center=0)
plt.title("Correlation Matrix")
plt.show()
```
